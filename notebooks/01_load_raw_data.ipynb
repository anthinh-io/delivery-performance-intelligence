# Đọc thử dữ liệu thô Olist

Đọc từng file CSV trong `data/raw/` bằng pandas, xác nhận không lỗi encoding/parsing.

```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../data/raw")
csv_files = sorted(RAW_DIR.glob("*.csv"))
print(f"Tìm thấy {len(csv_files)} file CSV:")
for f in csv_files:
    print(" -", f.name)
```

```python
dataframes = {}
errors = []

for f in csv_files:
    try:
        df = pd.read_csv(f)
        dataframes[f.stem] = df
        print(f"OK  {f.name:45s} shape={df.shape}")
    except (UnicodeDecodeError, pd.errors.ParserError) as e:
        errors.append((f.name, e))
        print(f"LỖI {f.name}: {e}")

print()
print(f"Đọc thành công {len(dataframes)}/{len(csv_files)} file, {len(errors)} lỗi.")
```

```python
assert len(csv_files) == 9, "Thiếu file — cần đủ 9 bảng Olist"
assert not errors, "Có lỗi encoding/parsing, cần kiểm tra lại"
```
