# Kiểm tra giá trị thiếu (null) trong dữ liệu thô Olist

Với mỗi bảng trong `data/raw/`, tính % giá trị thiếu theo từng cột.

```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../data/raw")
csv_files = sorted(RAW_DIR.glob("*.csv"))

dataframes = {f.stem: pd.read_csv(f) for f in csv_files}
print(f"Đã đọc {len(dataframes)} bảng.")
```

```python
def missing_pct(df):
    pct = (df.isna().sum() / len(df) * 100).round(2)
    return pct[pct > 0].sort_values(ascending=False)

missing_report = {}
for name, df in dataframes.items():
    pct = missing_pct(df)
    missing_report[name] = pct
    print(f"=== {name} (n={len(df)}) ===")
    if pct.empty:
        print("Không có giá trị thiếu.")
    else:
        print(pct.to_string())
    print()
```

## Tổng hợp

Danh sách các bảng có ít nhất một cột chứa giá trị thiếu, kèm số cột bị ảnh hưởng.

```python
summary = pd.Series({name: len(pct) for name, pct in missing_report.items() if not pct.empty})
summary = summary.sort_values(ascending=False)
summary.rename("số cột có giá trị thiếu").to_frame()
```
